```python
import pandas as pd

# Создаем данные (Data Preparation)
data = {
    'name': ['Ivan', 'Daria', 'Oleg', 'Maria', 'Alex', 'Elena', 'Petr'],
    'age': [20, 22, 21, 19, 23, 20, 22],
    'group': ['A', 'B', 'A', 'C', 'B', 'A', 'C']
}
# Индексы как ID (labels)
ids = ['id_1', 'id_2', 'id_3', 'id_4', 'id_5', 'id_6', 'id_7']
df = pd.DataFrame(data, index=ids)

print("Наш датафрейм:")
print(df)
print("-" * 30)

# 1. Выбираем 5-го студента по порядку
# Помни: в Python считаем с нуля, значит 5-й — это индекс 4.
# Используем .iloc (integer location)
student_5 = df.iloc[4]
print(f"5-й студент в списке:\n{student_5}")
print("-" * 30)

# 2. Средний возраст по группам
# Группируем (split), выбираем колонку age (select) и считаем среднее (apply function)
mean_age = df.groupby('group')['age'].mean()
print(f"Средний возраст по группам:\n{mean_age}")
print("-" * 30)

# 3. Добавляем 1 год через .apply()
# Мы передаем анонимную функцию lambda, которая берет значение x и возвращает x + 1
df['age'] = df['age'].apply(lambda x: x + 1)
print("Датафрейм после обновления возраста (всем +1 год):")
print(df.head())
```
